```python
# Getting category keywords for job title classification
category_map = {
    'Technology & IT': [
        'python', 'developer', 'programmer', 'software', 'database', 
        'systems', 'back-end', 'django', 'applications', 'administrator'
    ],
    'Healthcare & Clinical': [
        'medical', 'surgeon', 'nurse', 'psychiatrist', 'optician', 
        'ophthalmologist', 'therapist', 'audiological', 'cytogeneticist', 
        'radiographer', 'diagnostic'
    ],
    'Engineering & Hard Sciences': [
        'engineer', 'engineering', 'scientist', 'chemist', 'meteorologist', 
        'seismologist', 'ecologist', 'physicist', 'materials', 'structural',
        'automotive', 'manufacturing'
    ],
    'Media, Creative & Design': [
        'designer', 'producer', 'journalist', 'photographer', 'writer', 
        'editor', 'production', 'multimedia', 'graphic', 'printmaker', 
        'broadcast', 'broadcasting', 'television', 'radio', 'film', 'video'
    ],
    'Finance, Legal & Business': [
        'trader', 'banker', 'legal', 'barrister', 'insurance', 'broker', 
        'executive', 'underwriter', 'conveyancer', 'equities', 'bonds', 
        'futures', 'strategy', 'sales', 'loss adjuster'
    ],
    'Education & Public Service': [
        'teacher', 'officer', 'librarian', 'warden', 'ranger', 
        'immigration', 'inspector', 'conservation', 'academic', 'school'
    ],
    'Management & Operations': [
        'manager', 'adviser', 'planner', 'chief', 'assistant', 
        'editorial', 'cabin crew', 'tour', 'fitness'
    ]
}
```

```python
#Function to classify job titles into categories based on keywords
def classify_job_title(job_title):
    # Iterate through the category map and check if any of the keywords are present in the job title
    for category, keywords in category_map.items():
        for keyword in keywords:
            if keyword.lower() in job_title.lower():
                return category
    return 'Other'
```

```python
#Apply the classification function to the 'Job Title' column and create a new 'Industry' column
df['Industry'] = df['Job Title'].apply(classify_job_title)
print(df['Industry'].value_counts())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.ticker import MaxNLocator


# Set the style for the plot
sns.set_style('whitegrid')
plt.figure(figsize=(10, 5))

# Use value_counts() to get the totals per industry, and plot them
counts = df['Industry'].value_counts()
bars = counts.plot(kind='barh', color=sns.color_palette("mako", len(counts)))

# Ensure the x-axis has integer ticks
plt.gca().xaxis.set_major_locator(MaxNLocator(integer=True))

# Aesthetics: Customize the plot appearance
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['bottom'].set_color('#cccccc') # Make bottom line subtle
plt.gca().spines['left'].set_color('#cccccc')
# Add value labels to the bars
plt.bar_label(bars.containers[0], padding=5, fontweight='bold', color='#333333')

# Add labels and title
plt.title('Job Market Distribution by Industry Sector', fontsize=16, fontweight='bold', pad=20, color='#111111')
plt.xlabel('Total Number of Postings Found', fontsize=12, labelpad=10, color='#555555')
plt.ylabel('', fontsize=12)
plt.tight_layout()

# Save the plot as an image file
plt.savefig('scraped_market_data.png', dpi=300, bbox_inches='tight')
plt.show()
```
